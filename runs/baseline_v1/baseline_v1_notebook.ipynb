### **Experiment: Baseline Model (baseline_v1)**

* This experiment establishes the baseline configuration for a GPT-style decoder-only transformer trained from scratch on Wiki dataset.

* The goal of this experiment is to verify that the training pipeline is stable and that the model is able to learn basic language patterns 

* This baseline serves as a reference point for some future experiments, where individual components of the model or training process will be modified to study their impact on training behavior, generalization, and output quality.

### **Important Configs**
* to check full config details navigate to configs/baseline_v1.py

* **Model capacity**
    * d_model = 128
    * num_heads = 4
    * num_layers = 4

* **Training core**
    * learning_rate = 3e-4
    * batch_size = 16

### **Loss Curve**

```python
import pandas as pd
import matplotlib.pyplot as plt

file_path = r"metrics.csv"
df = pd.read_csv(file_path)

steps = df["step"]
train_loss = df["train_loss"]
val_loss = df["val_loss"]

plt.figure(figsize=(8, 5))  # wider graph
plt.plot(steps, train_loss, label="Train Loss")
plt.plot(steps, val_loss, label="Validation Loss")

plt.xlabel("Steps")
plt.ylabel("Loss")
plt.title("Training vs Validation Loss")

plt.legend()
plt.grid(True)

plt.savefig(r"loss_curve.png")
plt.show()
```

### **Observations**

**1. Training Behavior**

- Loss decreases smoothly over steps → stable training
- No major spikes observed → gradients are controlled
- Loss plateaus around ~6.2 after ~3000 steps → limited model capacity

**2. Generalization (Train vs Validation)**

- Train and validation loss are very close → good generalization
- No significant gap → no overfitting observed
- Both curves follow similar trend → consistent learning

**3. Output Behavior**

- Outputs follow basic English sentence structure
- Semantic meaning is weak or incorrect
- Some repetition and unnatural word combinations present
- Model captures syntax but not factual understanding
