### **Experiment2: Effect of Model Depth (Number of Layers)**

* In this experiment, the depth of the transformer model is increased to study its impact on learning behavior and performance.

* The baseline model uses 4 layers, and this experiment increases the number of layers to 8 while keeping all other configurations unchanged.

* The goal is to observe how increasing model depth affects training stability, convergence, generalization, and output quality in a small-scale setup.

* To check full config details navigate to configs/exp2_v1_layers.py

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load all three experiments
df_l8 = pd.read_csv("metrics.csv")
df_l4 = pd.read_csv("../baseline_v1/metrics.csv")   
fig, axes = plt.subplots(1, 2, figsize=(15, 5), sharey=True)

# layers = 8
axes[0].plot(df_l8["step"], df_l8["train_loss"], label="Train")
axes[0].plot(df_l8["step"], df_l8["val_loss"], linestyle="--", label="Val")
axes[0].set_title("Layers = 8")
axes[0].set_xlabel("Steps")
axes[0].set_ylabel("Loss")
axes[0].legend()
axes[0].grid(True)

# layers = 4 (baseline)
axes[1].plot(df_l4["step"], df_l4["train_loss"], label="Train")
axes[1].plot(df_l4["step"], df_l4["val_loss"], linestyle="--", label="Val")
axes[1].set_title("Layers = 4 (Baseline)")
axes[1].set_xlabel("Steps")
axes[1].legend()
axes[1].grid(True) 
```

### **Observations**

### **1. Training Behavior**

- Training remains stable even with 8 layers
- Convergence is slightly slower compared to baseline (4 layers)
- Final training loss is lower (~6.05 vs ~6.17 baseline)
- Gradients are slightly higher but remain controlled → no instability

### **2. Generalization (Train vs Validation)**
- Validation loss closely follows training loss
- Slight improvement in validation loss (~6.23 vs ~6.30 baseline)
- No strong signs of overfitting despite increased depth
- Generalization improves marginally with increased capacity

### **3. Output Behavior**
- Improvement is present but subtle, not dramatic

### **Conclusion**
* Increasing model depth from 4 to 8 layers improves performance slightly, leading to lower training and validation loss and marginally better outputs. However, this comes at the cost of slower training.

* This suggests that model depth contributes to learning capacity even in small-scale setups, but the benefits are incremental rather than significant.
